# ML-03 — Frame Your Lane as an ML Task

## 1. My lane as an ML task (type)
- **Task Type:** Binary Classification (Content Refresh Risk)
- **Why:** The goal is to predict whether a webpage requires a content refresh/update (1) or if its performance is stable (0). Binary classification provides a clear, actionable threshold for content teams to prioritize high-risk pages.

## 2. Target or proxy
- **Target Variable:** `needs_refresh` (Binary Label: 0 or 1)
- **Source:** Defined proxy target based on observed performance metrics. `1` represents pages with high impressions but declining CTR/clicks, while `0` represents stable pages.

## 3. Success metric
- **Primary Metric:** Precision@50 and F1-Score.
- **Defensible Benchmark:** Precision@50 >= 0.70. Out of the top 50 pages flagged for refresh, at least 35 should be actual high-risk pages.

## 4. The unit of analysis, as a real dataframe
- **Unit of Analysis:** 1 Row = 1 Webpage / Unique URL Slice.

```python
import pandas as pd
import numpy as np

# 1. Dataset Load Karein (Correct GitHub Raw Link)
url = 'https://raw.githubusercontent.com/CodewithSaira/ML-Pipelining/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(url)

# 2. Proxy Target Column Banayein (using impressions_90d & clicks_90d)
df['needs_refresh'] = np.where(
    (df['impressions_90d'] > df['impressions_90d'].median()) &
    (df['clicks_90d'] < df['clicks_90d'].median()),
    1, 0
)

# 3. Unit of Analysis Display Karein (Top 5 Rows)
cols_to_show = [col for col in ['impressions_90d', 'clicks_90d', 'ctr', 'avg_position', 'needs_refresh'] if col in df.columns]
df[cols_to_show].head()
```

## 5. Why ML beats a fixed rule here
- **Why ML Beats Rules:** Fixed `if/else` rules fail because high-volume head keywords behave differently than long-tail ones. ML captures non-linear interactions across decay, rank fluctuations, and click dynamics simultaneously without manual threshold tuning.
