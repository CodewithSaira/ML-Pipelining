## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*


### Trustworthy Action Queue
This section turns raw model numbers into an easy-to-read to-do list for humans. Each page is ranked by priority and given a simple reason code to explain why it needs attention.

### Simple Action Definitions
* **Improve Metadata (CTR Optimization) [`HIGH_IMPRESSION_LOW_CTR`]:** Many people see this page on Google, but very few click it. Fix it by rewriting the page title and description to get more clicks.
* **Content Refresh Required [`DECLINING_TRAFFIC_HIGH_VALUE`]:** This used to be a popular page, but it is slowly losing visitors over time. Fix it by updating old facts and adding fresh information.
* **Protect & Monitor [`STABLE_HIGH_PERFORMER`]:** This page is doing great and bringing in steady traffic. Avoid making big edits, and just keep an eye on its performance.
* **Evaluate for Consolidation / Pruning [`LOW_ENGAGEMENT_LOW_RANK`]:** Very few people are finding or reading this page. Either combine its useful parts into a better article or redirect/remove it.

```python
import os
import pandas as pd
import numpy as np

# Ensure required directories exist
os.makedirs("../outputs", exist_ok=True)
os.makedirs("../figures", exist_ok=True)

# Generate reproducible sample dataset reflecting search performance metrics
np.random.seed(42)
n_pages = 50

data = {
    "page_id": [f"page_{i:03d}" for i in range(1, n_pages + 1)],
    "impressions": np.random.randint(1500, 60000, size=n_pages),
    "ctr": np.random.uniform(0.008, 0.075, size=n_pages),
    "traffic_decay_rate": np.random.uniform(-0.35, 0.15, size=n_pages),
    "model_score": np.random.uniform(0.20, 0.98, size=n_pages)
}

df_playbook = pd.DataFrame(data)

# Map human-readable action types and reason codes
def assign_action_and_reason(row):
    if row["impressions"] > 25000 and row["ctr"] < 0.025:
        return "Improve Metadata (CTR Optimization)", "HIGH_IMPRESSION_LOW_CTR"
    elif row["traffic_decay_rate"] < -0.10:
        return "Content Refresh Required", "DECLINING_TRAFFIC_HIGH_VALUE"
    elif row["model_score"] > 0.70:
        return "Protect & Monitor", "STABLE_HIGH_PERFORMER"
    else:
        return "Evaluate for Consolidation / Pruning", "LOW_ENGAGEMENT_LOW_RANK"

results = df_playbook.apply(assign_action_and_reason, axis=1)
df_playbook["recommended_action"] = [res[0] for res in results]
df_playbook["reason_code"] = [res[1] for res in results]

# Rank queue by priority (Model Score descending, Impressions descending)
df_ranked_queue = df_playbook.sort_values(
    by=["model_score", "impressions"],
    ascending=[False, False]
).reset_index(drop=True)

# Display top ranked queue entries
print("--- Top 10 Ranked Content Action Queue ---")
print(df_ranked_queue[["page_id", "model_score", "reason_code", "recommended_action"]].head(10))

# Export queue output to work/outputs/
output_path = "../outputs/ranked_action_queue.csv"
df_ranked_queue.to_csv(output_path, index=False)
print(f"\nAction queue successfully exported to: {output_path}")
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*



### Who Uses This & For What
* **Target Users:** Content creators, SEO strategists, and editorial teams.
* **Purpose:** To quickly see which search pages need attention and prioritize work based on expected traffic impact.
* **Decision Support:** This tool provides recommendations to guide human decisions; it does not make changes automatically.

### Where This Tool Stops Being Valid
* **Not Causal:** The model shows patterns and correlations, not guaranteed cause-and-effect results.
* **Static Snapshot:** It relies on past data windows and cannot predict sudden Google algorithm updates or sudden market changes.
* **Search Only:** It only looks at search performance metrics (clicks, impressions, position). It does not review technical site bugs, brand tone, or content quality.

```python
# Section 2 Code: Define boundaries and check model limits

# Define boundary limits in a clear structure
model_limits = {
    "intended_users": "SEO & Content Editorial Teams",
    "primary_use_case": "Decision support for page refresh & metadata optimization",
    "automation_level": "Human-in-the-loop (No direct auto-publishing)",
    "data_dependencies": "Historical search performance metrics only",
    "known_invalid_cases": [
        "Sudden Google core algorithm updates",
        "Brand-new pages with zero search history",
        "Off-page technical site errors or server outages"
    ]
}

# Print system boundary check for verification
print("--- System Intended Use & Scope Check ---")
for key, value in model_limits.items():
    if isinstance(value, list):
        print(f"\n{key.replace('_', ' ').title()}:")
        for item in value:
            print(f"  - {item}")
    else:
        print(f"{key.replace('_', ' ').title()}: {value}")
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*


### What a Person Must Check Before Acting
Before taking any action recommended by the model, an editor or SEO strategist must verify:
* **Search Intent Alignment:** Check if the page still matches what users are actually searching for on Google.
* **Content Quality & Accuracy:** Ensure facts, dates, and brand voice remain accurate after any updates.
* **Technical Integrity:** Verify canonical tags and internal links before making major structural changes.

### The No-Go List (Actions Never Automated)
The following actions must NEVER be run on autopilot:
1. **Automated Content Deletion:** Pages must never be deleted or de-indexed automatically.
2. **Mass URL Redirects (301):** Combining or redirecting URLs always requires manual editor approval.
3. **Core Brand & Conversion Pages:** High-value landing pages are strictly excluded from automated edits.
4. **Metadata Overwrites:** Title tags and descriptions must be reviewed by a human to prevent brand voice errors.

```python
# Section 3 Code: Apply Human Review Policies and No-Go Rules

def apply_review_policy(row):
    # Rule 1: High-impact actions are flagged for mandatory human approval
    if row["recommended_action"] in ["Evaluate for Consolidation / Pruning", "Content Refresh Required"]:
        return "MANDATORY_HUMAN_REVIEW"

    # Rule 2: High-impression pages are high-risk assets
    if row["impressions"] > 30000:
        return "SENIOR_SEO_REVIEW"

    return "STANDARD_EDITORIAL_REVIEW"

# Apply policy to our ranked action queue
df_ranked_queue["review_policy"] = df_ranked_queue.apply(apply_review_policy, axis=1)

# Flag pages that fall under the No-Go safety check
df_ranked_queue["no_go_flag"] = df_ranked_queue["recommended_action"].apply(
    lambda x: "STRICT_NO_AUTO_EXECUTION" if x in ["Evaluate for Consolidation / Pruning", "Content Refresh Required"] else "SAFE_FOR_REVIEW"
)

# Display pages requiring strict human oversight
print("--- Human Review & Safety Gate Preview ---")
print(df_ranked_queue[["page_id", "reason_code", "recommended_action", "review_policy", "no_go_flag"]].head(10))
```

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*



### What Signals Stale Recommendations
Our recommendations become stale when real-world performance strays from model assumptions. We monitor three primary signals:

* **CTR Prediction Drift:** When actual click-through rates diverge significantly from expected model scores after metadata updates.
* **Traffic Decay Acceleration:** When pages categorized as stable experience sudden, unexpected drops in search impressions.
* **Algorithm & Seasonal Shifts:** When major Google search core updates or seasonality render historical performance data outdated.

### Retrain Triggers
* **Schedule-Based Trigger:** Retrain model quarterly (every 90 days) using fresh search performance data.
* **Performance-Based Trigger:** Retrain immediately if model accuracy drops by more than 15% over a 30-day window.
* **Data Drift Trigger:** Retrain when more than 20% of catalog pages show significant ranking position changes following a search engine core update.

```python
# Section 4 Code: Simulate Monitoring Metrics and Retrain Trigger Checks

# Define performance threshold boundaries
DRIFT_THRESHOLD = 0.15  # 15% performance drop threshold
DAYS_SINCE_LAST_TRAIN = 95  # Simulated age of current model in days

# Simulate recent model accuracy tracking
np.random.seed(10)
baseline_accuracy = 0.82
current_accuracy = np.random.uniform(0.60, 0.85)
accuracy_drop = (baseline_accuracy - current_accuracy) / baseline_accuracy

# Evaluate retrain triggers
triggers = {
    "schedule_trigger": DAYS_SINCE_LAST_TRAIN > 90,
    "performance_trigger": accuracy_drop > DRIFT_THRESHOLD
}

print("--- Monitoring & Retrain Status ---")
print(f"Days Since Last Model Retrain: {DAYS_SINCE_LAST_TRAIN} days")
print(f"Baseline Accuracy: {baseline_accuracy:.2%}")
print(f"Current Monitored Accuracy: {current_accuracy:.2%}")
print(f"Accuracy Drop: {accuracy_drop:.2%}")

if any(triggers.values()):
    print("\n[ACTION REQUIRED]: Retrain Trigger Fired!")
    if triggers["schedule_trigger"]:
        print(" -> Reason: Model schedule limit exceeded (> 90 days).")
    if triggers["performance_trigger"]:
        print(" -> Reason: Model accuracy drift exceeded allowable threshold (> 15%).")
else:
    print("\n[STATUS]: Model status healthy. No retrain required at this time.")
```

## 5. Exports for the paper

*Write the queue (and any figures you want to reuse) to work/outputs/ — your paper builds on these files.*


### Artifact & Figure Generation
This section exports our final ranked queue, summary metrics, and visual plots to `work/outputs/` and `work/figures/`. These generated files serve as the direct inputs and visual receipts for our research paper recommendations section.

### Export Deliverables
* **Ranked Queue CSV:** `work/outputs/ranked_action_queue.csv` — Contains complete list of prioritized pages, reason codes, and human review flags.
* **Summary Metrics JSON:** `work/outputs/playbook_summary_metrics.json` — Stores breakdown counts by action type.
* **Action Distribution Plot:** `work/figures/action_distribution.png` — Visual bar chart showing how pages are split across action types.

```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt

# 1. Ensure target export directories exist
os.makedirs("../outputs", exist_ok=True)
os.makedirs("../figures", exist_ok=True)

# 2. Export final ranked queue CSV to work/outputs/
queue_output_path = "../outputs/ranked_action_queue.csv"
df_ranked_queue.to_csv(queue_output_path, index=False)
print(f"[EXPORT]: Saved ranked action queue to: {queue_output_path}")

# 3. Export summary metrics JSON to work/outputs/
summary_counts = df_ranked_queue["recommended_action"].value_counts().to_dict()
metrics_payload = {
    "total_pages_evaluated": len(df_ranked_queue),
    "action_breakdown": summary_counts,
    "mandatory_human_review_count": int((df_ranked_queue["review_policy"] == "MANDATORY_HUMAN_REVIEW").sum())
}

metrics_path = "../outputs/playbook_summary_metrics.json"
with open(metrics_path, "w") as f:
    json.dump(metrics_payload, f, indent=4)
print(f"[EXPORT]: Saved playbook summary metrics to: {metrics_path}")

# 4. Generate and export action distribution plot to work/figures/
plt.figure(figsize=(9, 5))
df_ranked_queue["recommended_action"].value_counts().plot(kind="barh", color="#2b5c8f")
plt.title("Content Action Recommendations Breakdown")
plt.xlabel("Number of Pages")
plt.ylabel("Recommended Action")
plt.tight_layout()

figure_path = "../figures/action_distribution.png"
plt.savefig(figure_path, dpi=300)
plt.close()
print(f"[EXPORT]: Saved action distribution plot to: {figure_path}")
```
